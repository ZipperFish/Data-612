
## Mitigating the Harm of Recommender Systems

The articles show that recommender systems often cause problems because they are built to keep us engaged for as long as possible. The Wired piece explains that platforms like YouTube and Facebook end up pushing extreme content simply because it gets more clicks. The algorithms don't know what is good or bad, they just learn from what people watch and click on. This creates a cycle where users are shown more and more extreme material over time, whether it is conspiracy theories, hate speech, or false information. The Berkeley paper adds to this by showing that when people see an average rating before giving their own, they tend to follow the crowd. This makes ratings less honest and less diverse, which can mess up the recommendations even further.

Both articles offer ways to fix these issues. The Wired piece suggests that platforms should stop promoting harmful topics and use quality checks to filter out untrustworthy content. It also says that platforms need to be more open about how their algorithms work so that people can hold them accountable. The Berkeley study takes a more data-driven approach. They let users rate things twice, once before and once after seeing the average, and then used machine learning to correct the bias. This reduced the bias by over 76%, showing that these problems can be fixed if companies are willing to try. The FICO example from the first paper makes a similar point, showing that adjusting rules to give everyone a fair chance can improve fairness without losing too much profit. In the end, we already have the tools to make things better. What we need is the willingness to use them.
